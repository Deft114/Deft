## 实验2：函数、条件、循环与递归

```python
# 任务1：成绩等级函数
def score_level(score):
    if score < 0 or score > 100:
        raise ValueError(f"成绩 {score} 超出 0~100 范围")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

for s in [95, 85, 70, 55, 105]:
    try:
        print(f"score_level({s}) = {score_level(s)}")
    except ValueError as e:
        print(f"score_level({s}) 引发 ValueError: {e}")
```

```python
# 任务2：for 和 while 计算 1 到 100 的整数和
# for 循环
sum_for = 0
for i in range(1, 101):
    sum_for += i
print(f"for 循环结果: {sum_for}")

# while 循环
sum_while = 0
n = 1
while n <= 100:
    sum_while += n
    n += 1
print(f"while 循环结果: {sum_while}")

print(f"两种方法结果相同: {sum_for == sum_while}")
```

```python
# 任务3：递归版和迭代版阶乘
def factorial_rec(n):
    """递归版阶乘"""
    if n == 0:
        return 1
    return n * factorial_rec(n - 1)

def factorial_iter(n):
    """迭代版阶乘"""
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

for n in [0, 1, 5, 10]:
    r1 = factorial_rec(n)
    r2 = factorial_iter(n)
    print(f"{n}! = 递归:{r1}, 迭代:{r2}, 一致:{r1 == r2}")
```

```python
# 任务4：summary 函数
def summary(values):
    """返回 (最小值, 最大值, 平均值)"""
    return min(values), max(values), sum(values) / len(values)

data = [23, 45, 67, 12, 89, 34]
min_val, max_val, avg_val = summary(data)  # 元组解包
print(f"数据: {data}")
print(f"最小值: {min_val}, 最大值: {max_val}, 平均值: {avg_val:.2f}")
```

```python
# 任务5：alias 与 copy 的比较（使用单独演示列表）
demo_scores = [85, 58, 92, 76, 60, 89]
alias = demo_scores
copied = demo_scores.copy()

print("修改前:")
print(f"  demo_scores = {demo_scores}")
print(f"  alias     = {alias}")
print(f"  copied    = {copied}")

alias[0] = 100  # 修改 alias 中的元素

print("\n修改 alias[0] = 100 后:")
print(f"  demo_scores = {demo_scores}  # 原列表也被改变")
print(f"  alias     = {alias}")
print(f"  copied    = {copied}  # 副本不受影响")
print("\n结论：alias 与原列表指向同一对象，修改会互相影响；copy() 产生独立副本。")
```

## 实验4：文件读写、异常处理与模块

```python
# 任务1：工作目录与路径构造
print("当前工作目录:", os.getcwd())
data_dir = os.path.join(os.getcwd(), "exp01_data")
os.makedirs(data_dir, exist_ok=True)
print("数据目录:", data_dir)
```

```python
# 任务2：写入和读取文件
file_path = os.path.join(data_dir, "students.txt")

content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""

with open(file_path, "w", encoding="utf-8") as f:
    f.write(content.strip() + "\n")
print(f"文件已写入: {file_path}")

# 读取文件
with open(file_path, "r", encoding="utf-8") as f:
    lines = f.readlines()
print("文件内容:")
for line in lines:
    print(" ", line.rstrip())
```

```python
# 任务3：parse_student 函数
def parse_student(line):
    """解析一行学生数据，返回 (name, score) 元组"""
    line = line.strip()
    parts = line.split(",")
    if len(parts) != 2:
        raise ValueError(f"字段数量错误: 期望2个字段，实际{len(parts)}个")
    name = parts[0].strip()
    if not name:
        raise ValueError("姓名为空")
    try:
        score = int(parts[1].strip())
    except ValueError:
        raise ValueError(f"成绩不是有效整数: '{parts[1].strip()}'")
    if score < 0 or score > 100:
        raise ValueError(f"成绩 {score} 超出 0~100 范围")
    return (name, score)
```

```python
# 任务4：按行处理文件，捕获异常
valid_records = []
invalid_records = []

with open(file_path, "r", encoding="utf-8") as f:
    header = f.readline()  # 跳过表头
    for line_num, line in enumerate(f, start=2):
        line = line.strip()
        if not line:
            continue
        try:
            name, score = parse_student(line)
            valid_records.append({"name": name, "score": score})
        except ValueError as e:
            invalid_records.append((line_num, str(e)))
            print(f"第 {line_num} 行无效: {e}")

print(f"\n有效记录数: {len(valid_records)}")
print(f"无效记录数: {len(invalid_records)}")
print("有效记录:", valid_records)
```

```python
# 任务5：写入报告文件，演示 w 与 a 模式
report_path = os.path.join(data_dir, "report.txt")

valid_count = len(valid_records)
invalid_count = len(invalid_records)
avg = sum(r["score"] for r in valid_records) / valid_count
pass_rate = sum(1 for r in valid_records if r["score"] >= 60) / valid_count

# w 模式：写入（覆盖已有内容）
with open(report_path, "w", encoding="utf-8") as f:
    f.write(f"有效记录数: {valid_count}\n")
    f.write(f"无效记录数: {invalid_count}\n")
    f.write(f"平均分: {avg:.2f}\n")
    f.write(f"及格率: {pass_rate:.2%}\n")
print("w 模式写入报告完成（覆盖已有内容）")

with open(report_path, "r", encoding="utf-8") as f:
    print("报告内容:")
    print(f.read())

# a 模式：追加（在已有内容后追加）
with open(report_path, "a", encoding="utf-8") as f:
    f.write("\n--- 追加内容 ---\n")
    f.write("追加说明：a 模式不会覆盖原有内容\n")
print("a 模式追加完成（保留原有内容）")

with open(report_path, "r", encoding="utf-8") as f:
    print("报告内容:")
    print(f.read())
```

## 实验5：类、继承与对象复制

```python
# 任务1：定义 Student 类
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score
    
    def is_passed(self):
        return self.score >= 60
    
    def __str__(self):
        return f"Student({self.name}, {self.score})"

# 测试
s1 = Student("Alice", 85)
print(s1)
print(f"是否及格: {s1.is_passed()}")
```

```python
# 任务2：为有效记录创建 Student 对象
students = []
for record in valid_records:
    stu = Student(record["name"], record["score"])
    students.append(stu)
    print(f"{stu}  及格: {stu.is_passed()}")
```

```python
# 任务3：GraduateStudent 继承
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic
    
    def __str__(self):
        return f"GraduateStudent({self.name}, {self.score}, 研究方向: {self.research_topic})"

gs = GraduateStudent("Alice", 92, "Machine Learning")
print(gs)
print(f"是否及格: {gs.is_passed()}")
```

```python
# 任务4：对象复制比较
import copy

def compare_copy_methods():
    # 每次从新建的原始对象开始
    original = {"name": "Alice", "scores": [85, 90]}
    
    # 直接赋值
    assigned = original
    # copy.copy() 浅拷贝
    shallow = copy.copy(original)
    # copy.deepcopy() 深拷贝
    deep = copy.deepcopy(original)
    
    print("修改前:")
    print(f"  original = {original}")
    print(f"  assigned = {assigned}")
    print(f"  shallow  = {shallow}")
    print(f"  deep     = {deep}")
    
    # 修改内部列表
    original["scores"].append(95)
    
    print("\n修改 original['scores'].append(95) 后:")
    print(f"  original = {original}  # 被修改")
    print(f"  assigned = {assigned}  # 同一对象，也被修改")
    print(f"  shallow  = {shallow}  # 浅拷贝，内部列表是同一引用，也被修改")
    print(f"  deep     = {deep}  # 深拷贝，完全独立，不受影响")

compare_copy_methods()
```
