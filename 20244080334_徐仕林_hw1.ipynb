## 欧几里得算法

```python
# 1. gcd 函数
def gcd(m, n):
    m, n = abs(m), abs(n)
    while n != 0:
        m, n = n, m % n
    return m

# 2. 评估GCD
print(f"gcd(2018, 2019) = {gcd(2018, 2019)}")
print(f"gcd(1200, 300)  = {gcd(1200, 300)}")
print(f"gcd(5040, 60)   = {gcd(5040, 60)}")

# 3. 负数测试
print(f"gcd(-2018, 2019) = {gcd(-2018, 2019)}")
print(f"gcd(1200, -300)  = {gcd(1200, -300)}")
print(f"gcd(-1200, -300) = {gcd(-1200, -300)}")
print("负数的GCD与正数相同，因为GCD定义为能同时整除两数的最大正整数，取绝对值后结果一致。")
```

## 近似欧拉数 e

```python
# 1. euler_limit: e = lim(x->inf) (1 + 1/x)^x
def euler_limit(x):
    return (1 + 1 / x) ** x

print(f"euler_limit(100)   = {euler_limit(100)}")
print(f"euler_limit(10000) = {euler_limit(10000)}")
print(f"math.exp(1)        = {math.exp(1)}")
```

```python
# 2. euler_infinite_sum: 前 n 项之和 (1-indexed)
def euler_infinite_sum(n):
    total = 0.0
    factorial = 1
    for k in range(n):
        if k > 0:
            factorial *= k
        total += 1.0 / factorial
    return total

print(f"euler_infinite_sum(0) = {euler_infinite_sum(0)}")
print(f"euler_infinite_sum(1) = {euler_infinite_sum(1)}")
print(f"euler_infinite_sum(2) = {euler_infinite_sum(2)}")
print(f"euler_infinite_sum(3) = {euler_infinite_sum(3)}")
print(f"euler_infinite_sum(4) = {euler_infinite_sum(4)}")
print(f"euler_infinite_sum(10)= {euler_infinite_sum(10)}")
print(f"math.exp(1)           = {math.exp(1)}")
```

```python
# 3. euler_approx: 使用 while 循环找到使近似值在 epsilon 内的最小项数
def euler_approx(epsilon):
    true_e = math.exp(1)
    total = 0.0
    factorial = 1
    k = 0
    while abs(total - true_e) >= epsilon:
        if k > 0:
            factorial *= k
        total += 1.0 / factorial
        k += 1
    return total

print(f"euler_approx(0.1)   = {euler_approx(0.1)}")
print(f"euler_approx(0.01)  = {euler_approx(0.01)}")
print(f"euler_approx(0.001) = {euler_approx(0.001)}")
print(f"math.exp(1)         = {math.exp(1)}")
```

```python
# 4. print_euler_sum_table
def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(f"euler_infinite_sum({k}) = {euler_infinite_sum(k)}")

print_euler_sum_table(10)
```

## 测试整数属性

```python
# 1. is_power_of_2
def is_power_of_2(n):
    while n > 1:
        if n % 2 != 0:
            return False
        n = n // 2
    return n == 1

for num in [1, 2, 3, 4, 8, 16, 32, 64, 100, 128, 256, 1024]:
    print(f"is_power_of_2({num}) = {is_power_of_2(num)}")
```

```python
# 2. is_power: 判断 n 是否是 b 的幂
def is_power(b, n):
    if n == 1:
        return True
    if b == 1:
        return n == 1
    while n > 1:
        if n % b != 0:
            return False
        n = n // b
    return n == 1

print(f"is_power(2, 8)    = {is_power(2, 8)}")
print(f"is_power(2, 10)   = {is_power(2, 10)}")
print(f"is_power(3, 27)   = {is_power(3, 27)}")
print(f"is_power(3, 10)   = {is_power(3, 10)}")
print(f"is_power(5, 1)    = {is_power(5, 1)}")
print(f"is_power(5, 125)  = {is_power(5, 125)}")
print(f"is_power(10, 1000)= {is_power(10, 1000)}")
```
